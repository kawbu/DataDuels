```python
# Align rows by the smaller dataset length and build the new dataframe
n = min(len(set1), len(set2))
s1 = set1.iloc[:n].reset_index(drop=True)
s2 = set2.iloc[:n].reset_index(drop=True)
new_set = pd.DataFrame({
    'slug': s2['TitleSlug'].astype(str),
    'title': s2['questionTitle'].astype(str),
    'description': s2['content'].astype(str),
    'difficulty': pd.Categorical(s1['difficulty']),
    'topic_Tags': s2['topicTags'].astype(str),
    'starter_code': s2['codeDefinition'],
    'test_cases': s1['input_output']
})
# Show a preview
new_set.head()
```
